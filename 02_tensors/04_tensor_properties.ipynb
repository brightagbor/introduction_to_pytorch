### Gradient and Autograd Properties
*These properties are particularly improtant when training neural networks.*

```python
x = torch.randn(2, 4, 5)
print(x)
```

```python
# Check whether gradient tracking is requested for the tensor
x.requires_grad
```

```python
# Check if gradient is accumulated for a tensor when applicable
x.grad
```

```python
# Check the function that produced a tensor in the Autograd graph, if applicable
x.grad_fn
```

```python
# Check whether the tensor is a leaf in the Autograd graph
x.is_leaf
```

```python
# Checks whether a non-leaf tensor is configured to retain its gradient, where exposed
x.retains_grad
```

### Real Autograd example

```python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2
print(x)
print(y)
```

```python
print(f"Requires Gradients: {x.requires_grad}")
print(f"Is Leaf: {x.is_leaf}")
print(f"Gradient Funtion: {x.grad_fn}")


print(f"\nRequires Gradients: {y.requires_grad}")
print(f"Is Leaf: {y.is_leaf}")
print(f"Gradient Funtion: {y.grad_fn}")

y.backward()

print(f"\nGradients: {x.grad}")
```

```python
# Gradient of a non-leaf tensor

x = torch.tensor(2.0, requires_grad=True)
y = x.to(torch.float64)
z = y ** 2

print(x)
print(y)
print(z)
```

```python
y.retain_grad()
```

```python
print(y)
```

```python
z.backward()
```

```python
print(z)
```

```python
print(f"x.grad: {x.grad}")
print(f"y.grad: {y.grad}")
```

```python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2

print(x.requires_grad)
print(x.is_leaf)
print(x.grad_fn)

print(y.requires_grad)
print(y.is_leaf)
print(y.grad_fn)

y.backward()

print(x.grad)
```
