```python
# Verify if the tensor is normally distributed
def normal_distribution(tensor):


    print(f"Min: {tensor.min().item():.4f}")
    print(f"Max: {tensor.max().item():.4f}")
    print(f"Mean: {tensor.mean().item():.4f}")
    print(f"Std: {tensor.std().item():.4f}\n")

    # Heatmap
    plt.figure(figsize=(9, 7))
    # Convert tensor to a Numpy array before plotting
    sns.heatmap(tensor.numpy(), annot=True, cmap="Blues", vmin=0, vmax=1)
    plt.title("Tensor Heatmap")
    plt.show()

    # Histogram Plot
    plt.figure(figsize=(8, 5))
    # Use sns.histplot with kde=True
    sns.histplot(
        tensor.detach().numpy().flatten(),
        kde=True,
        bins=10,
        binrange=(0, 1),
        edgecolor="black",
    )
    plt.title("Value Distribution Histogram with KDE")
    plt.xlabel("Tensor Value")
    plt.ylabel("Count / Density")
    plt.show()  
```

```python
#4x4 random floats between 0 and 1
rand_tensor = torch.rand(10, 10)
rand_tensor
```

```python
normal_distribution(rand_tensor)
```

```python
randn_tensor = torch.randn(10, 10)
randn_tensor
```

```python
normal_distribution(randn_tensor)
```
