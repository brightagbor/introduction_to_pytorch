### Broadcasting in PyTorch

*Will dividing matrix A by B work?*

```python
import torch
```

```python
A = torch.tensor([
    [30, 45, 60, 15],
    [5, 10, 30, 5],
    [30, 20, 20, 30]
])
(print(A))
```

```python
B = torch.sum(A, axis=1)
print(B)
```

```python
print(f"Shape of Matrix `A: {A.shape}")
print(f"Shape of Matrix `B`: {B.shape}")
```

```python
# Division: Divide Matrix `A` by Matrix `B`
div_result = A / B
print(div_result)
```

```python
B = torch.unsqueeze(B, 1)
print(B)
```

```python
print(f"Shape of Matrix `A: {A.shape}")
print(f"Shape of Matrix `B`: {B.shape}")
```

```python
# Division: Divide   `A` by Matrix `B`
div_result = A / B
print(div_result)
```
