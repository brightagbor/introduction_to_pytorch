### Creating tensors from existing data
**`torch.tensor()` creates a new tensor by copying the supplied data**

```python
# From a list
tensor1 = torch.tensor([1, 2, 3, 4])
print(tensor1)
```

```python
# From a nested list (2D tensor)
tensor2 = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])
print(tensor2)
```

```python
# Creating tensors and specifying data types
x = torch.tensor([1, 2, 3], dtype=torch.float32)
x
```

### Creating tensors using `torch.from_numpy()`

*An important difference is that the NumPy array and the tensor share the same underlying memory for supported NumPy data types.*

```python
# Creating the numpy array
np_array = np.array([10, 20, 30])
np_array
```

```python
# Creating the tensor from the array
x = torch.from_numpy(np_array)
x
```

```python
# Verify the first item in the array
np_array[0]
```

```python
# Verify the first item in the tensor
x[0]
```

```python
# Modify the first item in the array
np_array[0] = 100
np_array
```

```python
# Verify the items in the tensor
x
```

*The tensor reflects the change because both objects share memory.*

### Creating tensors with predefined values
*These methods are useful when you need a tensor filled with zeros, ones, or a particular constant.*

#

```python
# torch.zeros()
# This is useful for initialising counters, accumulators, and arrays of defauls values.

x = torch.zeros(2, 3)
print(x)
```

```python
x = torch.zeros(5, 10)
print(x)
```

```python
# torch.ones()
# This is useful when you need a tensor of ones for mthematical operations


x = torch.ones(2, 5, 3)
print(x)
```

```python
# toch.full()
# Fills every element with s specified value
# The first argument is the shape, and the second is the fill value

x = torch.full((4,3), 7)
print(x)
```

### Creating tensors with random values
*Random initialisation is espeially important in machine learning and deep learning.*

```python
# torch.rand()
# Generates random values from a uniform distribution between 0 and 1

x = torch.rand(2, 3)
print(x)
```

```python
# torch.randn()
# Generates random values from a standard normal distribution, with mean 0 and standard deviation of 1
# This method is commony used in neural network experiments
# Unlike torch.rand(), values can be negative or positive, and they are not restricted to the interval from 0 to 1.

x = torch.randn(2, 3)
print(x)
```

```python
# torch.randint()
# Generates random intergers within a specified range
x = torch.randint(0, 10, (2, 3))
print(x)
```

```python
x = torch.randint(0, 5, (5, 10))
print(x)
```

```python
# torch.randperm()
# Generates a random permutation of integers from 0 up to n - 1
# This is useful when shuffling dataset indices

x = torch.randperm(10)
print(x)
```

### Creating tensors with sequences and ranges

```python
# torch.arange()
# This is used to create a sequence of values using a start, end, and optional step.

x = torch.arange(0, 10, 2)
print(x)



# syntax: torch.arange(start, end, step)
```

```python
start = 5
stop = 100
step = 10

x = torch.arange(start, stop, step=step)
print(x)
```

```python
# torch.linspace()
# Create a specified number of evenly spaced values between two endpoints

x = torch.linspace(0, 10, steps=50)
x
```

### Creating tensors using identity and diagonal matrices

```python
# torch.eye()
# Create a two-dimensional identity matrix
# NB: An identity matrix has ones along its main diagonal and zeros elsewhere

x = torch.eye(5)
print(x)
```

```python
# torch.diag()
# Creates a diagonal matrix from a one-dimensional tensor

x = torch.diag(torch.tensor([1, 2, 3, 4]))
print(x)
```

### Creating tensors based on another tensor
*These methods are particularly useful when you have a tensor and want another tensor with the same shape*

```python
# Original tensor
x = torch.tensor([[1, 2, 3],
                 [4, 5, 6]])
print(x)
```

```python
# torch.zeros_like()
# Creates a tesor of zeros withthe same shape as x

y = torch.zeros_like(x)
print(y)
```

```python
# torh.ones_like()
# Creates a tensor of ones with the same shape as x

y = torch.ones_like(x)
print(y)
```

```python
# torch.full_like()
# Creates a tensor with the same shape as x, filled with a specified value

y = torch.full_like(x, 9)
print(y)
```

```python
# torch.rand_like()
# Creates random values with the same shape as x

y = torch.rand_like(x, dtype=torch.float32)
print(y)
```
