### Sum: `torch.sum()`
*`torch.sum()` adds tensor elements together.

```python
x = torch.tensor([10.0, 20.0, 30.0, 40.0])
sum_result = torch.sum(x)
print(sum_result)
```

*Summiing along a dimension. The key is understanding what dim means:* 
- dim=0: reduce the first dimension, adding orresponding rows together. The result contains columns sums.
- dim=1: reduce the second dimension, adding across each row. The result contains row sums.


```python
x = torch.tensor([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])

print(x)
```

```python
# Dimension of x
print(x.dim())
```

```python
# Display the shape of the tensor
x.shape
```

```python
sum_dim0 = torch.sum(x, dim=0)
print(sum_dim0)
```

```python
sum_dim1 = torch.sum(x, dim=1)
print(sum_dim1)
```

### Maximum and Minimum
*torch.max() and torch.min() identify extreme values*

```python
x = torch.tensor([12, 5, 27, 9, 18, 2.0, 1.5])
print(x)
```

```python
# Maximum value in tensor x
print(torch.max(x))
```

```python
# Minimum value in tensor x
print(torch.min(x))
```

```python
# Retrieving the maximum value and its index
values = x
max_value, max_index = torch.max(values, dim=0)
print(max_value, max_index)
```

```python
# Retrieving the minimum value and its index
values = x
min_value, min_index = torch.min(values, dim=0)
print(min_value, min_index)
```

```python
#### Retrieving max and min value for a 2D tensor
x = torch.tensor([
    [3, 8, 2, 4],
    [9, 1, 6, 10],
])
print(x)
```

```python
values, indices = torch.max(x, dim=1)
print(values, indices)
```

### Argmax and Argmin
*Sometimes you do not need the maximum value itself. You need its position.*
- torch.argmax() returns the index of the maximum value.
- torch.argmin() returns the index of the minimum value.

```python
x = torch.tensor([0.15, 0.70, 0.10, 0.05])
print(x)
```

```python
# Return the index of the maximum value
print(torch.argmax(x))
```

```python
# Return the index of the minimum value
print(torch.argmin(x))
```

```python
x2d = torch.tensor([
    [0.15, 0.70, 0.10, 0.05],
    [0.85, 0.95, 0.10, 0.11]
    ])
print(x)
```

### Standard Deviation and Varian
* These operations measure how spread out values are
1. `Variance` measures the average squared deviation from the mean, subject to the chosen statistical convention.
2. `Standard deviation` is the square root of variance.

```python
x = torch.tensor([2.0, 4.0, 6.0, 8.0])
print(x)
```

```python
# Print the variance
print(torch.var(x))
```

```python
# Print the standard deviation
print(torch.std(x))
```

### Note:
*By default, PyTorch uses the sample variance convention (correction=1), which divides by `n-1`. To alculate population variance instead, use the below:*

```python
print(torch.var(x, correction=0))
print(torch.std(x, correction=0))
```

***The distinction matters when you are analysisng a complete population versus estimating population variation from a sample.***

### The Keepdim parameter
*When aggregating along a dimension, keepdim=True preserves the reduced dimension as size 1*

```python
x = torch.tensor([
    [1.0, 2.0, 3.0, 4.0],
    [4., 5.0, 6.0, 11.0]
])

print(x)
```

```python
# Without keepdim=True
sum_x = torch.sum(x, dim=1)
print(f"Sum: {sum_x}")
print(f"Shape: {sum_x.shape}")
```

```python
# With keepdim=True
sum_x_keepdim = torch.sum(x, dim=1, keepdim=True)
print(f"Sum: {sum_x_keepdim}")
print(f"Shape: {sum_x_keepdim.shape}")
```

### Demonstrating keepdim=True

*Suppose each row represents a sample containing three features. You want to divide each element by its row's sum.*

```python
x
```

```python
row_sum = x.sum(dim=1, keepdim=True)
print(row_sum)
```

```python
print(x.shape)
```

```python
print(row_sum.shape)
```

```python
normalised = x/row_sum
print(normalised)
```

```python
print(normalised.shape)
```

```python
# repeat the above with keepdim=False
row_sum = x.sum(dim=1)
print(row_sum)
```

```python
normalised_new = x / row_sum
print(normalised_new)
```
