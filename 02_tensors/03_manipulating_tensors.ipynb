```python
import torch
```

### Adding and Removing Dimensions

```python
x = torch.tensor([1.0, 2.0, 3.0])
x
```

```python
# Add dim at position 0
x.unsqueeze(0)
```

```python
## Add dim at position 1
x.unsqueeze(1)
```

```python
# Add dim at last
x.unsqueeze(-1)
```

```python
# Add dim at last
x.unsqueeze(-1).unsqueeze(0)
```

```python
# shape: [1, 3, 1]
y = torch.rand(1, 3, 2)
y
```

```python
y.shape
```

```python
# remove all size-1 dims → [3]
y.squeeze()
```

```python
# remove dim 0 only → [3, 1]
y.squeeze(0)
```

```python
# remove last dim        → [1, 3]
y.squeeze(-1)
```

### Stacking and Concatenating

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
```

```python
# Stacking
torch.stack([a, b])
```

```python
torch.stack([a, b], dim=0)
```

```python
torch.stack([a, b], dim=1)
```

```python
# Concatenation
torch.cat([a, b])
```

```python
A = torch.rand(2, 3)
B = torch.rand(2, 3)
```

```python
A
```

```python
B
```

```python
# Stack rows
torch.cat([A, B], dim=0)
```

```python
# Stack columns
torch.cat([A, B], dim=1)
```

### Splitting and Chunking

```python
x = torch.arange(12)
x
```

```python
x = x.reshape(4, 3)
```

```python
# Split into chunks of 2 rows
torch.split(x, 2, dim=0)
```

```python
# Split into 2 equal chunks
torch.chunk(x, 2, dim=0)
```

```python
a, b = torch.split(x, 2, dim=0)
print(a)
print(b)
```

```python
# TYPE CASTING 
x = torch.tensor([1, 2, 3])                  # dtype: int64
```

```python
# Float32
print(x.float())
print(x.float().dtype)
```

```python
# Double
print(x.double())
print(x.double().dtype)
```

```python
print(x.half())
print(x.half().dtype)
```

```python
print(x.int())
print(x.int().dtype)
```

```python
x.int()
```

```python
x.bool()
```

```python
# Explicit casting
x.to(torch.float32)
```
