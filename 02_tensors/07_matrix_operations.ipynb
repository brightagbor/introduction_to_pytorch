### Element-wise multiplication

```python
a = torch.tensor([
    [1, 2],
    [3, 4]
])

b = torch.tensor([
    [5, 6],
    [7, 8]
])
print(a)
print(b)
```

```python
# Element-wise multiplication
print(a*b)
```

### Matrix Multiplication

```python
print(a)
```

```python
print(b)
```

```python
# Matrix Multiplication
print(a@b)
```

```python
# Matrix Multiplication
print(torch.matmul(a, b))
```

*NB: PyTorch supports both `torch.matmul()` and the `@` operator.*

```python
a = torch.tensor([[1.0, 2.0],
                  [3.0, 4.0]])

b = torch.tensor([[5.0, 6.0],
                  [7.0, 8.0]])

print(torch.matmul(a, b))
print(a @ b)
```

### torch.mm() versus torch.bmm()

```python
# torch.mm(A, B)
# torch.mm(A, B) performs matrix multplication on two two-dimensional tensors.

A = torch.randn(3, 4)
B = torch.randn(4, 2)
```

```python
print(A)
```

```python
print(B)
```

```python
print(A.shape, B.shape)
```

```python
print(A@B)
print((A@B).shape)
```

```python
print(torch.matmul(A, B))
print(torch.matmul(A, B).shape)
```

```python
print(torch.mm(A, B))
print((torch.mm(A, B).shape))
```

### torch.bmm(A, B)

*`torch.bmm(A, B)` performs batch matrix multiplication on two three-dimensional tensors, where the first dimension represents the batch.*

```python
A = torch.randn(4, 3, 4)
B = torch.randn(4, 4, 2)

print(torch.bmm(A, B))
```

### Dot Products
*The dot product multiplies corresponding elements of two vectors and adds the results*

```python
# Matrix a
a = torch.tensor([
    [1, 2],
    [3, 4]
])

b = torch.tensor([
    [5, 6],
    [7, 8]
])

print(a)
print(b)
```

```python
# Dot Produt of a and b
print(torch.dot(a, b))
```

```python
a = torch.tensor([1, 2])
b = torch.tensor([3, 4])

print(a, b)
```

```python
print(a.dot(b))
```

```python
print(a.matmul(b))
```

```python
print(a*b)
```

```python
print(torch.dot(a, b))
```

```python
print((torch.dot(a, b)).shape)
```

```python
A = torch.randn(3, 4)
print(A)
```

```python
B = torch.randn(5, 4)
print(B)
```

```python
A_matmul_B = A @ B
print(A_matmul_B)
```

```python
# Transposing B into matrix C
C = B.T
print(C)
```

```python
# Shape of B and C
print(f"Shape of Matrix B: {B.shape}")
print(f"Shape of Matrix C: {C.shape}")
```

```python
# Now the inner dimensions are equal after transposing
# We can now do matrix multiplication

D = A @ C
print(D)
```

```python
print(f"Shape of Matrix D (Result of A@C - A@B.T): {D.shape}")
```
