### Autograd Implementation in PyTorch

```python
import torch
```

```python
# if y= wx + b; given x=3, w=2, b=1
# Therefore, y = 2(3) + 1 = 7
# partial derivative of y w.r.t w=x=3
# and partial derivative of y w.r.t b=1

# Implementing the above with PyTorch
```

```python
# Define the input
x = torch.tensor(3.0)
```

```python
# Define trainiable parameters
# requires_grad=True tells PyTorch to track operations involving 
# the tensor so that gradients can be calculated.
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
```

```python
# Forward propagation
y = w*x + b
print(y)
```

```python
# Calculate gradients (using the backward pass)
y.backward()
print(y)
```

```python
print("Prediction: ", y.item())
print("Gradient of w: ", w.grad.item())
print("Gradient of b: ", b.grad.item())
```

### Example II

```python
def function(inputs):
    x, y = inputs

    output1 = x**2 + y
    output2 = 2*x + 3*y

    return torch.stack([output1, output2])
```

```python
# Input tensor
inputs = torch.tensor(
    [2.0, 3.0],
    requires_grad=True
)
```

```python
# Calculate Jacobian
jacobian = torch.autograd.functional.jacobian(
    function, inputs
)

print(jacobian)
```

### Understanding th egradient using PyTorch

```python
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)


# Forward computation
f = x**2 + 3*x*y + y**2

# Backward differentiation
f.backward()

print("Funtion value: ", f.item())
print("Gradient with respect to x: ", x.grad.item())
print("Gradient with respect to y: ", y.grad.item())
```
