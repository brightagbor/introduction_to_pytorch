```python
# Basic import for reproducibility
import os

# Configure CUDA environment variables
os.environ["CUBLAS_WORKSPACE_CONFIG"] = ":4096:8"

import numpy as np
import matplotlib.pyplot as plt
import random
import torch
import torch.nn as nn
```

```python
# Python basic reproducibility (lightweight)

# Random seed
SEED = 42

# Set random seed in the random module
random.seed(SEED)

# Set random seed in the numpy module
np.random.seed(SEED)

# Set random seed in PyTorch Library for CPU
torch.manual_seed(SEED)

# Set random seed in PyTorch CUDA 
torch.cuda.manual_seed(SEED) # for one GPU
torch.cuda.manual_seed_all(SEED) # for all other GPUs
```

```python
# function for reproducibility (lightweight)
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)

    if torch.cuda.is_available():
        # Set random seed in PyTorch CUDA 
        torch.cuda.manual_seed(SEED) # for one GPU
        torch.cuda.manual_seed_all(SEED) # for all other GPUs
    # Set random seed in PyTorch Library for CPU
    torch.manual_seed(SEED)
```

```python
# Enabling deterministic algorithms in Pytorch
torch.use_deterministic_algorithms(True)
```

```python
# Configuring cuDNN for deterministic operations
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
```

```python
# Reproduibility in data loading
# Create the dataset
from torch.utils.data import DataLoader
from torchvision import transforms, datasets


dataset = datasets.FakeData(
    size = 10000,
    image_size=(3, 32, 32),
    num_classes=10,
    transform=transforms.ToTensor()
)

# Data Loader with random seed
generator = torch.Generator()
generator.manual_seed(SEED)

loader = DataLoader(
    dataset = dataset,
    batch_size=32,
    shuffle=True,
    generator=generator
)
```

```python
# Reproducibility in data loading + number of workers seed
def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % (2**32)
    np.random.seed(worker_seed)
    random.seed(worker_seed)

loader = DataLoader(
    dataset = dataset,
    batch_size=32,
    shuffle=True,
    num_workers=4,
    worker_init_fn=seed_worker,
    generator=generator
)
```
