```python
for i in range(torch.cuda.device_count()):
    props = torch.cuda.get_device_properties(i)
    print(props)
```

```python
for i in range(torch.cuda.device_count()):
    props = torch.cuda.get_device_properties(i)
    
    print(f"\nGPU {i}: {props.name}")
    print(f"VRAM: {props.total_memory/1024**3:.1f} GiB")
    print(f"Compute Cap: {props.major}")
    print(f"Compute Cap: {props.major}.{props.minor}")
    print(f"SM Count: {props.multi_processor_count}")
```

```python
### Quick tensor test on GPU
if torch.cuda.is_available():
    x = torch.rand(3, 3).cuda()
    print(f"Tensor on GPU: {x.device}")
    print(f"Sample values: \n{x}")
else:
    print("GPU not available -- running on CPU only")
```

#### `x = torch.rand(3, 3).cuda()`
Creates a 3×3 matrix of random floats between 0 and 1, then moves it from CPU RAM to GPU VRAM via `.cuda()`.


#### `print(f"Tensor on GPU: {x.device}")`
Prints where the tensor lives — if GPU is working, you'll see cuda:0 (meaning GPU #0), not cpu.

```python
#### System Resources Check


import psutil

ram = psutil.virtual_memory()
print(f"Total RAM:     {ram.total / 1024**3:.1f} GB")
print(f"Available RAM: {ram.available / 1024**3:.1f} GB")
print(f"Used RAM:      {ram.used / 1024**3:.1f} GB")
print(f"RAM Usage:     {ram.percent}%")
```

```python
### Check for SWAP

import psutil

swap = psutil.swap_memory()
print(f"Total Swap:  {swap.total / 1024**3:.1f} GB")
print(f"Used Swap:   {swap.used / 1024**3:.1f} GB")
print(f"Free Swap:   {swap.free / 1024**3:.1f} GB")
print(f"Swap Usage:  {swap.percent}%")
```
