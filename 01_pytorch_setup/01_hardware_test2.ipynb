```python
import psutil


ram = psutil.virtual_memory()
print(f"Total RAM: {ram.total/1000**3:.2f}")
print(f"Available RAM: {ram.available/1000**3:.2f}")
print(f"Used Memory: {ram.used/1000**3:.2f}")
print(f"Used Percentage (%): {ram.percent}%")
print(f"Active Memmory: {ram.active/1000**3:.2f}")
print(f"Active Memory: {ram.active/1000**3:.2f}")
print(f"Free Memory: {ram.free/1000**3:.2f}")
```

```python
# Check SWAP Memory
import psutil

swap = psutil.swap_memory()
print(swap)
print(f"Total RAM: {swap.total/1000**3:.2f}")
print(f"Used SWAP Memory: {swap.used/1000**3:.2f}")
print(f"Free SWAP Memory: {swap.free/1000**3:.2f}")
print(f"Used SWAP Percentage (%): {swap.percent}%")
print(f"Sin : {swap.sin}")
print(f"Sout : {swap.sout}")
```
