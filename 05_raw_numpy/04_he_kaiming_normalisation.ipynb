```python
import numpy as np
```

### Implementing He Normal Initialisation

```python
fan_in = 5
fan_out = 5

std = np.sqrt(2/fan_in)
weights = np.random.normal(
    loc=0.0, scale=std, size=(fan_in, fan_out)
)
print(weights)
```

### Implementing He Uniform Initialisation

```python
limit = np.sqrt(6/(fan_in))

weights = np.random.uniform(
    low=-limit, high=limit, size=(fan_in, fan_out)
)
print(weights)
```
