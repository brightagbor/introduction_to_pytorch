```python
import numpy as np
```

```python
np.random.seed(42)
```

### Xavier Uniform Initialisation

```python
fan_in = fan_out = 5 # Layer input and output
print(fan_in)
print(fan_out)
```

```python
limit = np.sqrt(6/(fan_in + fan_out))
print(limit)
```

```python
weights = np.random.uniform(
    low=-limit, high=limit, size=(fan_in, fan_out)
)
print(weights)
```

### Xavier Normal Initialisation

```python
std = np.sqrt(2/(fan_in + fan_out))
weights = np.random.normal(
    loc=0.0, scale=std, size=(fan_in, fan_out)
)

print(weights)
```

```python
# Initialise weights and biases
w1 = np.random.randn(fan_in, fan_out) * np.sqrt(2/fan_out)
print("Weights")
print(w1)
```
