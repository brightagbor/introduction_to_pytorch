### STEP 1: Preparing and Loading Data
*To showcase this, let's create some known data using the linear regression formula. We'll use a linear regression formula to make a straigt line with known `parameters`*

```python
# Create `known` parameters
# We want to build a model that will be able to estimate the values below

weight = 0.7
bias = 0.3
```

```python
# Create
start = 0
end = 1
step = 0.02

X = torch.arange(start=start, end=end, step=step).unsqueeze(dim=1)
y = weight*X + bias
```

```python
print(X.shape, y.shape)
```

```python
### Splitting data into training and test split
train_split = int(0.8 * len(X))
X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]
```

```python
# Verify the lengths of the splits
print(len(X_train))
print(len(y_train))
print(len(X_test))
print(len(y_test))
```

### Visualising the Data

```python
def plot_predictions(train_data, 
                     train_labels,
                     test_data,
                     test_labels,
                     predictions=None):
    """
    Plots training data, test data and compares predictions.
    """
    plt.figure(figsize=(10, 7))

    # Plot training data in blue
    plt.scatter(train_data, train_labels, c="b", s=6, label="Training Data")

    # Plot training data in red
    plt.scatter(test_data, test_labels, c="g", s=6, label="Testing Data")

    # Check for predictions
    if predictions is not None:
        # Plot the predictions f they exist
        plt.scatter(test_data, predictions, c="r", s=6, label="Predictions")

    # SHow the legend
    plt.legend(prop={"size": 14})
```

```python
plot_predictions(X_train, y_train, X_test,y_test)
```
